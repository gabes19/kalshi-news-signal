## Data Cleaning

```python
# Clean dataset
if 'df' not in globals() or not isinstance(df, pd.DataFrame):
    df = pd.read_csv('data/joined_data.csv')

clean_df = df.copy()
clean_df['date'] = pd.to_datetime(clean_df['date'], errors='coerce')
clean_df = (
    clean_df.dropna(subset=['date'])
    .sort_values('date')
    .drop_duplicates(subset=['date'])
    .reset_index(drop=True)
)

drop_all_missing = [c for c in clean_df.columns if clean_df[c].isna().all()]
drop_high_missing = [c for c in clean_df.columns if clean_df[c].isna().mean() > 0.90]
clean_df = clean_df.drop(columns=sorted(set(drop_all_missing + drop_high_missing)))

if {'sentiment_neg_pct', 'sentiment_pos_pct', 'headline_count'}.issubset(clean_df.columns):
    denom = clean_df['headline_count'].replace(0, np.nan)
    clean_df['sentiment_neg_rate'] = (clean_df['sentiment_neg_pct'] / denom).clip(0, 1).fillna(0)
    clean_df['sentiment_pos_rate'] = (clean_df['sentiment_pos_pct'] / denom).clip(0, 1).fillna(0)

if 'unique_sources' in clean_df.columns:
    clean_df = clean_df.drop(columns=['unique_sources'])

market_cols = [c for c in clean_df.columns if c.startswith('cpi')]
if market_cols:
    clean_df[market_cols] = clean_df[market_cols].ffill().bfill()

print(f'clean_df shape: {clean_df.shape}')
clean_df.head()
```

```python
# Market-specific sub-dataframes
if 'clean_df' not in globals() or not isinstance(clean_df, pd.DataFrame):
    raise ValueError('Run the Clean dataset cell first to create clean_df.')

base_cols = [
    c for c in clean_df.columns
    if not c.startswith('cpi_') and not c.startswith('cpi_core_') and not c.startswith('cpi_yoy_')
]

cpi_cols = [c for c in clean_df.columns if c.startswith('cpi_') and not c.startswith('cpi_core_') and not c.startswith('cpi_yoy_')]
cpi_core_cols = [c for c in clean_df.columns if c.startswith('cpi_core_')]
cpi_yoy_cols = [c for c in clean_df.columns if c.startswith('cpi_yoy_')]

cpi_df = clean_df[base_cols + cpi_cols].copy()
cpi_core_df = clean_df[base_cols + cpi_core_cols].copy()
cpi_yoy_df = clean_df[base_cols + cpi_yoy_cols].copy()

print('cpi_df:', cpi_df.shape)
print('cpi_core_df:', cpi_core_df.shape)
print('cpi_yoy_df:', cpi_yoy_df.shape)
cpi_df.head()
```

```python
# Bucket df - distribution uncertainty (entropy) vs headline count over time
if 'clean_bucket_df' not in globals() or not isinstance(clean_bucket_df, pd.DataFrame):
    raise ValueError('Run the Clean bucket dataset cell first to create clean_bucket_df.')

if 'headline_count' not in clean_bucket_df.columns or 'date' not in clean_bucket_df.columns:
    raise ValueError('Missing required columns: date/headline_count')

def market_bucket_entropy(df, market_prefix):
    bucket_cols = [
        c for c in df.columns
        if c.startswith(f'{market_prefix}_bucket_') and not c.endswith('bucket_total_probability')
    ]
    if not bucket_cols:
        return pd.Series(np.nan, index=df.index)

    p = df[bucket_cols].apply(pd.to_numeric, errors='coerce').fillna(0).clip(lower=0)
    row_sum = p.sum(axis=1).replace(0, np.nan)
    p = p.div(row_sum, axis=0)

    eps = 1e-12
    entropy = -(p * np.log(p + eps)).sum(axis=1)
    return entropy

plot_df = clean_bucket_df[['date', 'headline_count']].dropna(subset=['date']).sort_values('date').copy()
plot_df['cpi_bucket_entropy'] = market_bucket_entropy(clean_bucket_df, 'cpi')
plot_df['cpi_core_bucket_entropy'] = market_bucket_entropy(clean_bucket_df, 'cpi_core')
plot_df['cpi_yoy_bucket_entropy'] = market_bucket_entropy(clean_bucket_df, 'cpi_yoy')
plot_df = plot_df[plot_df['date'] > '2025-07-01']

fig, ax1 = plt.subplots(figsize=(14, 6))
ax2 = ax1.twinx()

ax1.plot(plot_df['date'], plot_df['cpi_bucket_entropy'], label='cpi bucket entropy', linewidth=2)
ax1.plot(plot_df['date'], plot_df['cpi_core_bucket_entropy'], label='cpi core bucket entropy', linewidth=2)
ax1.plot(plot_df['date'], plot_df['cpi_yoy_bucket_entropy'], label='cpi yoy bucket entropy', linewidth=2)

ax2.plot(plot_df['date'], plot_df['headline_count'], label='headline count', color='black', linestyle='--', alpha=0.7)

ax1.set_title('Bucket Distribution Uncertainty (Entropy) vs Headline Count')
ax1.set_xlabel('Date')
ax1.set_ylabel('Distribution uncertainty (entropy)')
ax2.set_ylabel('Headline count')

lines1, labels1 = ax1.get_legend_handles_labels()
lines2, labels2 = ax2.get_legend_handles_labels()
ax1.legend(lines1 + lines2, labels1 + labels2, loc='upper left')
plt.tight_layout()
plt.show()
```

## Correlation Analysis

```python
# Correlation analysis: sentiment/headline features vs implied volatility (clean_df and clean_bucket_df)
sentiment_features = ['headline_count', 'mean_sentiment', 'median_sentiment', 'sentiment_neg_rate', 'sentiment_pos_rate']
vol_features = ['cpi_implied_volatility', 'cpi_core_implied_volatility', 'cpi_yoy_implied_volatility']

def volatility_correlation_table(df, label):
    required = [c for c in sentiment_features + vol_features if c in df.columns]
    if not required:
        raise ValueError(f'{label}: no required columns found for correlation analysis')

    corr = df[required].corr(numeric_only=True)
    rows = [c for c in sentiment_features if c in corr.index]
    cols = [c for c in vol_features if c in corr.columns]
    table = corr.loc[rows, cols]
    return table

if 'clean_df' not in globals() or not isinstance(clean_df, pd.DataFrame):
    raise ValueError('Run the clean_df setup cells first.')
if 'clean_bucket_df' not in globals() or not isinstance(clean_bucket_df, pd.DataFrame):
    raise ValueError('Run the clean_bucket_df setup cells first.')

corr_clean_df = volatility_correlation_table(clean_df, 'clean_df')
corr_clean_bucket_df = volatility_correlation_table(clean_bucket_df, 'clean_bucket_df')

print('Correlation table - clean_df')
display(corr_clean_df.round(3))

print('Correlation table - clean_bucket_df')
display(corr_clean_bucket_df.round(3))

corr_diff = (corr_clean_bucket_df - corr_clean_df).abs()
print('Absolute difference (bucket - standard)')
display(corr_diff.round(6))
```

### Significant Findings
- The correlation patterns are effectively identical in `clean_df` and `clean_bucket_df` for implied-vol columns (differences are ~0), since both datasets share the same implied-volatility features after cleaning.
- Strongest relationship: `headline_count` vs `cpi_yoy_implied_volatility` is **positive** at about **0.260** (weak-to-moderate positive).
- Next strongest: `headline_count` vs `cpi_core_implied_volatility` is **negative** at about **-0.148** (weak negative).
- Sentiment-level correlations with implied volatility are generally small (mostly between about **-0.10 and 0.10**), suggesting limited linear signal from sentiment alone in this sample.

## Cyclical Analysis

```python
# Cyclical analysis: spikes, calendar effects, and lead/lag structure
if 'clean_df' not in globals() or not isinstance(clean_df, pd.DataFrame):
    raise ValueError('Run the clean_df setup cells first.')

vol_cols = ['cpi_implied_volatility', 'cpi_core_implied_volatility', 'cpi_yoy_implied_volatility']
missing_vol = [c for c in vol_cols if c not in clean_df.columns]
if missing_vol:
    raise ValueError(f'Missing volatility columns: {missing_vol}')

cyc_df = clean_df[['date', 'headline_count'] + vol_cols].dropna(subset=['date']).sort_values('date').copy()
cyc_df['vol_composite'] = cyc_df[vol_cols].mean(axis=1)
cyc_df['day_name'] = cyc_df['date'].dt.day_name()
cyc_df['month'] = cyc_df['date'].dt.to_period('M').astype(str)

headline_spike_threshold = cyc_df['headline_count'].quantile(0.90)
vol_spike_threshold = cyc_df['vol_composite'].quantile(0.90)
cyc_df['headline_spike'] = cyc_df['headline_count'] >= headline_spike_threshold
cyc_df['vol_spike'] = cyc_df['vol_composite'] >= vol_spike_threshold
cyc_df['both_spike'] = cyc_df['headline_spike'] & cyc_df['vol_spike']

dow_summary = (
    cyc_df.groupby('day_name')[['headline_count', 'vol_composite']]
    .mean()
    .reindex(['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday'])
)
month_summary = cyc_df.groupby('month')[['headline_count', 'vol_composite']].mean()

lag_corr_rows = []
for lag in range(-7, 8):
    lag_corr_rows.append({
        'lag_days': lag,
        'corr_headline_vs_vol_shifted': cyc_df['headline_count'].corr(cyc_df['vol_composite'].shift(lag))
    })
lag_corr_df = pd.DataFrame(lag_corr_rows).sort_values('lag_days').reset_index(drop=True)

print(f'Headline spike threshold (90th pct): {headline_spike_threshold:.3f}')
print(f'Volatility spike threshold (90th pct composite): {vol_spike_threshold:.3f}')
print(f'Headline spikes: {int(cyc_df["headline_spike"].sum())}')
print(f'Volatility spikes: {int(cyc_df["vol_spike"].sum())}')
print(f'Both spikes same day: {int(cyc_df["both_spike"].sum())}')

display(dow_summary.round(3))
display(month_summary.round(3))
display(lag_corr_df.round(3))
display(cyc_df.loc[cyc_df['both_spike'], ['date', 'headline_count', 'vol_composite'] + vol_cols])
```

```python
# Cyclical plots: spikes and calendar summaries
fig, axes = plt.subplots(3, 1, figsize=(14, 16))

# 1) Time series with spike markers
ax1 = axes[0]
ax1.plot(cyc_df['date'], cyc_df['vol_composite'], label='vol composite', color='tab:blue', linewidth=2)
ax1.scatter(
    cyc_df.loc[cyc_df['vol_spike'], 'date'],
    cyc_df.loc[cyc_df['vol_spike'], 'vol_composite'],
    color='tab:red', s=40, label='vol spike (>=90th pct)', zorder=3
)
ax1_t = ax1.twinx()
ax1_t.plot(cyc_df['date'], cyc_df['headline_count'], label='headline count', color='black', linestyle='--', alpha=0.6)
ax1_t.scatter(
    cyc_df.loc[cyc_df['headline_spike'], 'date'],
    cyc_df.loc[cyc_df['headline_spike'], 'headline_count'],
    color='tab:orange', s=30, label='headline spike (>=90th pct)', zorder=3
)
ax1.set_title('Volatility and Headline Spikes Over Time')
ax1.set_ylabel('Composite volatility')
ax1_t.set_ylabel('Headline count')
l1, lab1 = ax1.get_legend_handles_labels()
l2, lab2 = ax1_t.get_legend_handles_labels()
ax1.legend(l1 + l2, lab1 + lab2, loc='upper left')

# 2) Day-of-week cycle
ax2 = axes[1]
dow_plot = dow_summary.reset_index().rename(columns={'index': 'day_name'})
ax2.plot(dow_plot['day_name'], dow_plot['headline_count'], marker='o', label='headline_count')
ax2_t = ax2.twinx()
ax2_t.plot(dow_plot['day_name'], dow_plot['vol_composite'], marker='o', color='tab:green', label='vol_composite')
ax2.set_title('Average by Day of Week')
ax2.set_ylabel('Headline count')
ax2_t.set_ylabel('Composite volatility')
l1, lab1 = ax2.get_legend_handles_labels()
l2, lab2 = ax2_t.get_legend_handles_labels()
ax2.legend(l1 + l2, lab1 + lab2, loc='upper left')

# 3) Lead/lag correlation profile
ax3 = axes[2]
ax3.plot(lag_corr_df['lag_days'], lag_corr_df['corr_headline_vs_vol_shifted'], marker='o')
ax3.axhline(0, color='gray', linewidth=1, linestyle='--')
ax3.set_title('Lead/Lag Correlation: headline_count vs shifted vol_composite')
ax3.set_xlabel('Lag (days): corr(headline_t, vol_{t+lag})')
ax3.set_ylabel('Correlation')

plt.tight_layout()
plt.show()
```

### Key Findings (Cyclical Analysis)
- Using 90th-percentile thresholds, there are **25 headline spikes** and **25 volatility spikes**, but only **1 same-day overlap** (around **2025-09-12**), so extreme headline and volatility days rarely coincide exactly.
- Calendar pattern: headline volume is highest on **Fridays** and lowest on weekends, while volatility is relatively elevated on **weekends/Wednesday** versus Thursday; the weekly cycles are not aligned.
- Monthly pattern: headline activity ramps up in **Aug-Oct 2025** (peaking in October), while composite volatility is elevated in **January** and **September-November**, indicating partial but imperfect seasonal co-movement.
- Lead/lag check shows the strongest correlation near **lag -5 days** (about **r ~ 0.232**), suggesting volatility changes can lead headline intensity by several days more than they move contemporaneously (same-day **r ~ 0.194**).
- Interpretation: this is exploratory correlation on daily aggregates (no intraday time column in this dataset), so results are directional rather than causal.


```python
# News-cycle expansion: CPI/Fed/current-events vs spikes
import re
if 'clean_df' not in globals() or not isinstance(clean_df, pd.DataFrame):
    raise ValueError('Run the clean_df setup cells first.')

vol_cols = ['cpi_implied_volatility', 'cpi_core_implied_volatility', 'cpi_yoy_implied_volatility']
missing_vol = [c for c in vol_cols if c not in clean_df.columns]
if missing_vol:
    raise ValueError(f'Missing volatility columns: {missing_vol}')

# Build daily spike frame
news_cycle_df = clean_df[['date', 'headline_count'] + vol_cols].dropna(subset=['date']).sort_values('date').copy()
news_cycle_df['vol_composite'] = news_cycle_df[vol_cols].mean(axis=1)
news_cycle_df['headline_spike'] = news_cycle_df['headline_count'] >= news_cycle_df['headline_count'].quantile(0.90)
news_cycle_df['vol_spike'] = news_cycle_df['vol_composite'] >= news_cycle_df['vol_composite'].quantile(0.90)
news_cycle_df['both_spike'] = news_cycle_df['headline_spike'] & news_cycle_df['vol_spike']

# Load headline-level text for event-topic detection
headline_path_candidates = [
    '../data/headlines/all_headline_data.csv',
    'data/headlines/all_headline_data.csv',
]
headline_path = next((p for p in headline_path_candidates if Path(p).exists()), None)
if headline_path is None:
    raise ValueError('Could not find all_headline_data.csv in expected paths.')

raw_headlines = pd.read_csv(headline_path)
if 'date' not in raw_headlines.columns:
    raise ValueError('all_headline_data.csv is missing date column.')

raw_headlines['date'] = pd.to_datetime(raw_headlines['date'], errors='coerce').dt.tz_localize(None).dt.normalize()
text_cols = [c for c in ['title', 'description'] if c in raw_headlines.columns]
if not text_cols:
    raise ValueError('all_headline_data.csv is missing title/description columns.')

raw_headlines['text_blob'] = raw_headlines[text_cols].fillna('').agg(' '.join, axis=1).str.lower()

# Heuristic cycle dictionaries
cycle_terms = {
    'cpi_release_cycle': [
        'consumer price index', 'cpi', 'inflation report', 'bls', 'bureau of labor statistics',
    ],
    'fed_cycle': [
        'federal reserve', 'fed', 'fomc', 'powell', 'rate decision', 'dot plot', 'policy statement',
    ],
    'current_events_cycle': [
        'tariff', 'oil', 'energy prices', 'geopolitical', 'war', 'election', 'shutdown',
        'debt ceiling', 'jobs report', 'payrolls', 'unemployment',
    ],
}

for cycle_name, terms in cycle_terms.items():
    pattern = '|'.join([re.escape(t) for t in terms])
    raw_headlines[cycle_name] = raw_headlines['text_blob'].str.contains(pattern, regex=True, na=False).astype(int)

daily_cycle = raw_headlines.groupby('date', as_index=False).agg(
    article_count=('text_blob', 'size'),
    cpi_release_mentions=('cpi_release_cycle', 'sum'),
    fed_mentions=('fed_cycle', 'sum'),
    current_event_mentions=('current_events_cycle', 'sum'),
)
for col in ['cpi_release_mentions', 'fed_mentions', 'current_event_mentions']:
    daily_cycle[col + '_share'] = (daily_cycle[col] / daily_cycle['article_count'].replace(0, np.nan)).fillna(0)

news_cycle_df = news_cycle_df.merge(daily_cycle, on='date', how='left')
for col in ['article_count', 'cpi_release_mentions', 'fed_mentions', 'current_event_mentions',
            'cpi_release_mentions_share', 'fed_mentions_share', 'current_event_mentions_share']:
    if col in news_cycle_df.columns:
        news_cycle_df[col] = news_cycle_df[col].fillna(0)

# Simple calendar proxy for CPI release window (typically early-mid month)
news_cycle_df['cpi_window_proxy'] = news_cycle_df['date'].dt.day.between(8, 16).astype(int)

def compare_means(flag_col):
    on = news_cycle_df[news_cycle_df[flag_col] == 1]
    off = news_cycle_df[news_cycle_df[flag_col] == 0]
    return pd.Series({
        'days_flag_on': len(on),
        'headline_mean_on': on['headline_count'].mean(),
        'headline_mean_off': off['headline_count'].mean(),
        'vol_mean_on': on['vol_composite'].mean(),
        'vol_mean_off': off['vol_composite'].mean(),
        'headline_spike_rate_on': on['headline_spike'].mean(),
        'vol_spike_rate_on': on['vol_spike'].mean(),
    })

# Binary event flags by high-share days
share_flags = {
    'high_cpi_release_day': (news_cycle_df['cpi_release_mentions_share'] >= news_cycle_df['cpi_release_mentions_share'].quantile(0.75)).astype(int),
    'high_fed_day': (news_cycle_df['fed_mentions_share'] >= news_cycle_df['fed_mentions_share'].quantile(0.75)).astype(int),
    'high_current_event_day': (news_cycle_df['current_event_mentions_share'] >= news_cycle_df['current_event_mentions_share'].quantile(0.75)).astype(int),
}
for k, v in share_flags.items():
    news_cycle_df[k] = v

summary = pd.DataFrame({
    'cpi_window_proxy': compare_means('cpi_window_proxy'),
    'high_cpi_release_day': compare_means('high_cpi_release_day'),
    'high_fed_day': compare_means('high_fed_day'),
    'high_current_event_day': compare_means('high_current_event_day'),
}).T

corr_cols = ['headline_count', 'vol_composite', 'cpi_release_mentions_share', 'fed_mentions_share', 'current_event_mentions_share']
corr_table = news_cycle_df[corr_cols].corr(numeric_only=True).loc[['headline_count', 'vol_composite'], ['cpi_release_mentions_share', 'fed_mentions_share', 'current_event_mentions_share']]

# Show top spike days with dominant cycle tag
tmp = news_cycle_df.copy()
tmp['dominant_cycle'] = tmp[['cpi_release_mentions_share', 'fed_mentions_share', 'current_event_mentions_share']].idxmax(axis=1)
spike_table = tmp.loc[tmp['both_spike'] | tmp['headline_spike'] | tmp['vol_spike'], [
    'date', 'headline_count', 'vol_composite', 'headline_spike', 'vol_spike', 'both_spike',
    'cpi_release_mentions_share', 'fed_mentions_share', 'current_event_mentions_share', 'dominant_cycle'
]].sort_values(['both_spike', 'headline_count', 'vol_composite'], ascending=[False, False, False]).head(20)

display(summary.round(3))
display(corr_table.round(3))
display(spike_table)

print('Quick interpretation helpers:')
print('- Compare *_mean_on vs *_mean_off to see how cycle days differ from baseline.')
print('- Compare headline_spike_rate_on and vol_spike_rate_on across cycle flags.')
print('- Use dominant_cycle on spike days to inspect likely event drivers.')
```

### Significant Findings (Cyclical + News Cycles)
- Same-day co-spikes are rare: headline and volatility spikes overlap infrequently, so attention surges and pricing surges are often not perfectly synchronized.
- Weekly cycle mismatch: headline volume tends to cluster on weekdays (especially late week), while volatility patterns are flatter or shifted; this suggests different drivers for media flow vs market repricing.
- Seasonal ramp: both series rise into late summer/fall, but not at identical rates; October headline intensity can remain elevated even when volatility does not peak proportionally.
- Lead/lag signal is modest: the strongest relationship appears with a short lag window rather than same-day only, consistent with information digestion over multiple days.
- Event-cycle interpretation: CPI-release/Fed/current-event keyword days can coincide with spikes, but results are exploratory and should be treated as directional (keyword heuristics, daily aggregation, no intraday timestamp alignment).
